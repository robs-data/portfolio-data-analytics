```python
# ============================================================
# IMPORTAÇÃO DE BIBLIOTECA
# ============================================================

import pandas as pd


# ============================================================
# CARREGAMENTO DOS DADOS (EXTRACT)
# ============================================================

# Caminho do arquivo Excel
arquivo = "../dados_bruto/EVENTOS.xlsx"

# Leitura das planilhas
events = pd.read_excel(arquivo, sheet_name="events")
emp = pd.read_excel(arquivo, sheet_name="EMP")
cat = pd.read_excel(arquivo, sheet_name="Cat_Events")
```

```python
# Visualizar dados brutos
display(events)
display(emp)
display(cat)
```

```python
# ============================================================
# EXPLORAÇÃO INICIAL DOS DADOS (EDA)
# ============================================================

# Ver estrutura das tabelas (tipos, nulos, colunas)
display(events.info())
display(emp.info())
display(cat.info())
```

```python
# ============================================================
# TRATAMENTO DOS DADOS (TRANSFORM)
# ============================================================

# ---------- 1. Converter coluna de data ----------
# Transformar a coluna de string para formato datetime
events["data_evnt"] = pd.to_datetime(events["data_evnt"])
```

```python
# ---------- 2. Padronizar nomes das colunas ----------
# Renomear colunas para facilitar análise e uso em SQL/Power BI
events.columns = [
    "data_evento",
    "id_empresa",
    "tipo_evento",
    "pagamento",
    "total_participantes",
    "cobertura",
    "coordenador",
    "valor_faturado",
    "royalties",
    "desconto_contratante",
    "contratante",
    "categoria"
]

emp.columns = [
"id_empresa",
"empresa",
"total_funcionarios"
]

cat.columns = [
"id_categoria",
"categoria",
"visao"
]
# Conferir estrutura após ajuste
# ============================================================
display(events.info())
display(emp.info())
display(cat.info())
```

```python
# ---------- 3. Tratamento de valores nulos ----------
# Verificar quantidade de valores nulos na coluna de faturamento
events["valor_faturado"].isnull().sum()

# Criar coluna indicando se o evento é gratuito
# True = evento sem faturamento (gratuito)
# False = evento com faturamento (pago)
events["evento_gratuito"] = events["valor_faturado"].isnull()

# Substituir valores nulos por 0 para permitir cálculos
events["valor_faturado"] = events["valor_faturado"].fillna(0)


# ============================================================
# RESULTADO FINAL (LOAD)
# ============================================================

# Visualizar base tratada
display(events)
```

```python
events.to_csv("../dados_tratados/events.csv", index=False)
emp.to_csv("../dados_tratados/empresas.csv", index=False)
cat.to_csv("../dados_tratados/categorias.csv", index=False)
```
